```python
import numpy as np

marks = np.array([40, 55, 62, 88, 95, 33, 71, 58])

# 1. 
increased_marks = np.round(marks * 1.15, 2)
print(increased_marks)

# 2. 
corrected_marks = np.minimum(increased_marks, 100)
print(corrected_marks)

# 3. 
gain = corrected_marks - marks
count_less_than_10 = np.sum(gain < 10)
print(count_less_than_10)

# 4.
failing_before = np.sum(marks < 40)


failing_after = np.sum(corrected_marks < 40)

print(failing_before)
print(failing_after)

# 5. 
topper_marks = np.max(corrected_marks)
percentage = np.round((corrected_marks / topper_marks) * 100, 1)
print(percentage)

# 6. 
total_marks_given = np.sum(corrected_marks - marks)
print(total_marks_given)
```

```python
import numpy as np
marks = np.array([72, 45, 88, 95, 30, 67, 91, 54, 78, 39])

# 1. 
average = np.mean(marks)
median = np.median(marks)

print("Average:", average)
print("Median:", median)


# 2. 
gap = np.max(marks) - np.min(marks)
print("Gap:", gap)


# 3.
topper_index = np.argmax(marks)
weakest_index = np.argmin(marks)

print("Topper index:", topper_index)
print("Weakest index:", weakest_index)


# 4. 
sorted_marks = np.sort(marks)
second_highest = sorted_marks[-2]
print("Second highest:", second_highest)


# 5. 
above_average = np.sum(marks > average)
print("Above average:", above_average)


# 6.
below_median = np.sum(marks < median)
print("Below median:", below_median)


# 7.
marks_without_topper = np.delete(marks, topper_index)
new_average = np.mean(marks_without_topper)

print("Average after removing topper:", round(new_average, 2))
```

```python
import numpy as np

marks = np.array([
    [72, 85, 60, 91],
    [45, 30, 98, 55],
    [88, 66, 74, 100],
    [95, 92, 55, 40],
    [30, 48, 62, 78]
])


# 1. 
subject_average = np.mean(marks, axis=0)

print("Subject averages:", subject_average)

student_total = np.sum(marks, axis=1)

print("Student totals:", student_total)


# 2.
toughest_subject = np.argmin(subject_average)

print("Toughest subject index:", toughest_subject)


# 3. 
top_student = np.argmax(student_total)

print("Top student index:", top_student)


# 4.
top_marks_each_subject = np.max(marks, axis=0)

print("Top marks in each subject:", top_marks_each_subject)


# 5.
students_above_60 = np.sum(marks > 60, axis=0)

print("Students above 60:", students_above_60)


# 6.
subject_range = np.max(marks, axis=0) - np.min(marks, axis=0)

most_spread_subject = np.argmax(subject_range)

print("Most spread out subject index:", most_spread_subject)


# 7. 
chemistry_marks = marks[:, 2]

lowest_chemistry_student = np.argmin(chemistry_marks)

total_of_lowest_chemistry_student = student_total[lowest_chemistry_student]

print("Total marks:", total_of_lowest_chemistry_student)
```

```python
import numpy as np

positions = np.array([0, 5, 12, 20, 30])

distances = np.abs(positions[:, None] - positions[None, :])
distances = distances.astype(float)
np.fill_diagonal(distances, np.inf)

print(distances)

farthest_distance = np.max(distances, axis=1)
print(farthest_distance)

nearest_to_zero = np.argmin(distances[0, 1:]) + 1
print(nearest_to_zero)

upper = distances.copy()
np.fill_diagonal(upper, np.inf)
upper[np.tril_indices(5)] = np.inf
closest_pair = np.unravel_index(np.argmin(upper), upper.shape)
print(closest_pair)

roll_numbers = np.arange(1, 25).reshape(-1, 6)
print(roll_numbers)
```
