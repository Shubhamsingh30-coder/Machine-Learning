```python
prices = " 120, 45 , 300 ,80, 95 "

price_list = []

for price in prices.split(","):
    price_list.append(int(price.strip()))

total = sum(price_list)
highest = max(price_list)
average = total / len(price_list)

print(f"Total : {total}")
print(f"Highest : {highest}")
print(f"Average : {average:.2f}")
```

```python
sentence = "Python is easy and powerful"

words = sentence.split()

total_words = len(words)

vowel_count = 0

for word in words:
    if word[-1].lower() in "aeiou":
        vowel_count += 1

longest_word = ""

for word in words:
    if len(word) > len(longest_word):
        longest_word = word

print(f"Total words : {total_words}")
print(f"Words ending with vowel : {vowel_count}")
print(f"Longest word : {longest_word}")
```

```python
sentence = "python is easy python is powerful python"

words = sentence.lower().split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

sorted_words = sorted(frequency.items(), key=lambda x: x[1], reverse=True)

print(frequency)
print(f"Most frequent word : {sorted_words[0][0]}")
print(f"Frequency : {sorted_words[0][1]}")
```

```python
sales = {
    "mon": 200,
    "tue": 350,
    "wed": 200
}

inverted = {}

for day, value in sales.items():
    if value in inverted:
        inverted[value].append(day)
    else:
        inverted[value] = [day]

print(inverted)
```

```python
test1 = {
    "amit": 40,
    "bina": 35
}

test2 = {
    "amit": 30,
    "chirag": 45
}

result = {}

for student, marks in test1.items():
    result[student] = marks

for student, marks in test2.items():
    if student in result:
        result[student] += marks
    else:
        result[student] = marks

print(result)
```

```python
employees = [
    ("Ravi", 55000),
    ("Sita", 72000),
    ("Amit", 48000)
]

sorted_employees = sorted(
    employees,
    key=lambda employee: employee[1],
    reverse=True
)

for employee in sorted_employees:
    print(f"{employee[0]} earns Rs. {employee[1]}")
```

```python
movies = [
    {"title": "Movie A", "rating": 8.2, "genre": "Thriller"},
    {"title": "Movie B", "rating": 8.4, "genre": "Sports"},
    {"title": "Movie C", "rating": 8.1, "genre": "Comedy"},
    {"title": "Movie D", "rating": 8.2, "genre": "Drama"}
]

print("Movies with rating > 8.1:")

for movie in movies:
    if movie["rating"] > 8.1:
        print(movie["title"])

genres = []

for movie in movies:
    if movie["genre"] not in genres:
        genres.append(movie["genre"])

print("\nGenres:")
print(genres)

rating_groups = {}

for movie in movies:
    rating = movie["rating"]
    title = movie["title"]

    if rating in rating_groups:
        rating_groups[rating].append(title)
    else:
        rating_groups[rating] = [title]

print("\nMovies grouped by rating:")

for rating, titles in rating_groups.items():
    print(f"Rating {rating}: {', '.join(titles)}")
```
