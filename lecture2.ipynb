```python
marks = "45, 88,72, 90,  60  "
parts = marks.split(",")

ml = []
for i in parts:
    ml.append(int(i))   

total = 0
for i in ml:
    total += i

avg = total / len(ml)

print(ml)
print(total)
print(avg)
```

```python
name = "  rahul KUMAR Sharma "
name = name.strip().title()
lst = name.split()

valid = True

for i in lst:
    if not i.isalpha():
        valid = False

if valid:
    ans = " ".join(lst)
    print(ans)
else:
    print("INVALID NAME")
```
